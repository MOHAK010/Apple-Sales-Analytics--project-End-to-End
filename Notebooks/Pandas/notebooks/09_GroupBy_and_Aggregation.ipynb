```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

Calculate the Total Quantity Sold for every Product.

```python
product_sales = (sales
                 .groupby("product_id")
                 .agg(
                     total_quantity = ("quantity", "sum")
                      )
                      .reset_index()
                      )

product_sales
```

Count the total number of products available in each Category.

```python
category_products = (
    products.groupby("category_id")
    .agg(
        total_products = ("product_id", "count")
    )
    .reset_index()
)

category_products
```

Find the total number of sales handled by every Store.

```python
store_sales = (sales
               .groupby("store_id")
               .agg(total_sales = ("sale_id", "count")
                    )
                    .reset_index()
               )

store_sales
```

Calculate the Average Quantity Sold by each Country.

```python
country_sales = (
    sales.merge(stores, on = "store_id")
    .groupby("country")
    .agg(
        avg_sales = ("quantity" , "mean")
         )
         .reset_index()
         )
print(country_sales)
```

Prepare a Store Performance Report containing Store Name, Country, Total Sales Records, Total
Quantity Sold, and Average Quantity Sold.

```python
store_performance = (
    sales.merge(stores,on = "store_id")
                    .groupby(["store_name", "country"])
                    .agg(
                        total_sales = ("sale_id","count"),
                         total_quantity_Sold = ("quantity","sum"),
                         average_quantity_sold =("quantity","mean")
                         )
                         .reset_index()
                         )
store_performance
```
