```python
import pandas as pd
import numpy as np 
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

```python
Convert Country, Category, Warranty Status, and Product Name into Category dtype wherever
appropriate.
```

```python
master_sales = sales.merge(products, on="product_id")
master_sales = master_sales.merge(category, on="category_id")
master_sales = master_sales.merge(stores, on="store_id")
master_sales = master_sales.merge(warranty, on="sale_id")
```

```python
cols = ["country", "category_name", "claim_status", "product_name"]

master_sales[cols] = master_sales[cols].astype("category")
```

```python
print(master_sales.dtypes)
```

```python
Compare the memory usage before and after optimization.
```

```python
master_sales.memory_usage(deep=True)  # before optimization
```

```python
master_sales.memory_usage(deep=True)  # After optimization
```

```python
Find the Top 10 Categories by Product Count.
```

```python
top_10_categories = (
    master_sales.groupby("category_name")
    .agg(total_products = ("product_name", "count"))
    .sort_values(by = "total_products", ascending=False)
    .head(10)
    )
print(top_10_categories)
```

```python
Create a MultiIndex using Country and Category, then generate a summary report.
```

```python
master_sales["revenue"] = master_sales["quantity"] * master_sales["price"]
```

```python
summary_report = (
    master_sales
    .groupby(["country", "category_name"])
    .agg(
        total_revenue=("revenue", "sum"),
        total_quantity=("quantity", "sum"),
        average_revenue=("revenue", "mean"),
        total_sales=("sale_id", "count")
    )
)

print(summary_report)
```

Optimize the Master Sales DataFrame for maximum performance while preserving all business
information and document every optimization applied.

```python
master_sales["sale_date"] = pd.to_datetime(master_sales["sale_date"])
master_sales["launch_date"] = pd.to_datetime(master_sales["launch_date"])
```

```python
cols = [
    "country",
    "category_name",
    "claim_status",
    "product_name"
]

master_sales[cols] = master_sales[cols].astype("category")
```

```python
master_sales.memory_usage(deep=True)
```
