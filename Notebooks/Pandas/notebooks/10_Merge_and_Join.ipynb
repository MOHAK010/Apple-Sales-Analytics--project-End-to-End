```python
import pandas as pd 
import numpy as np 
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

Merge the Sales dataset with the Products dataset using Product ID.

```python
merge_dataset = products.merge(sales, on = "product_id")
merge_dataset
```

Merge Store information into the Sales dataset.

```python
merge_dataset = sales.merge(
    stores,
    on="store_id"
)

print(merge_dataset)
```

Merge Category information into the Products dataset.

```python
merge_dataset = products.merge(
    category,
    on="category_id"
)

print(merge_dataset)
```

Merge Warranty information with the Sales dataset using Sale ID

```python
merge_dataset = sales.merge(
    warranty,
    on="sale_id"
)

print(merge_dataset)
```

Create a final Master Sales DataFrame containing Sale Date, Product Name, Category, Store Name,
Country, Quantity, Price, and Warranty Status. This DataFrame will be used throughout the remaining
project and later in Power BI.

```python
master_sales = (
    sales
    .merge(products, on="product_id")
    .merge(category, on="category_id")
    .merge(stores, on="store_id")
    .merge(warranty, on="sale_id")
)

master_sales = master_sales[
    [
        "sale_date",
        "product_name",
        "category_name",
        "store_name",
        "country",
        "quantity",
        "price",
        "claim_status"
    ]
]

print(master_sales)
```
