```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

Sort all products based on Price in descending order

```python
products.sort_values(by="price", ascending=False)
```

Display the Top 10 most expensive Apple products

```python
print(products.nlargest(10,"price")[["product_name", "price"]])
```

```python
print(products.nsmallest(10,"price")[["product_name","price"]])
```

Assign a Price Rank to every product.

```python
products["price"] = products["price"].rank( ascending = False)
print(products)
```

Take a random sample of 500 sales records for quick inspection.

```python
print(sales.sample(500))
```

Prepare a Premium Product Report showing Product Name, Category, Launch Date, Price, and Price
Rank

```python
premium_report = products.merge(
    category,
    on="category_id"
)

```

```python
premium_report["price_rank"] = premium_report["price"].rank(
    ascending=False)
```

```python
premium_report = premium_report[
    [
        "product_name",
        "category_name",
        "launch_date",
        "price",
        "price_rank"
    ]
].sort_values(
    by="price",
    ascending=False
)

```

```python
print(premium_report)
```
