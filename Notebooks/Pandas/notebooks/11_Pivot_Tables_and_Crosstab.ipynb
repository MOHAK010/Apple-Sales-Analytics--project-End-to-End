```python
import pandas as pd
import numpy as np 
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

Create a Pivot Table showing Total Revenue by Country

```python
master_sales = sales.merge(products, on="product_id")
master_sales = master_sales.merge(category, on="category_id")
master_sales = master_sales.merge(stores, on="store_id")
master_sales = master_sales.merge(warranty, on="sale_id")
```

```python
master_sales["revenue"] = (
    master_sales["quantity"] * master_sales["price"]
)
```

```python
pivot = master_sales.pivot_table(
    index="country",
    values="revenue",
    aggfunc="sum"
)

print(pivot)
```

Create a Pivot Table showing Total Quantity Sold by Category.

```python
pivot_table1 = master_sales.pivot_table(
    index = "category_name",
    values = "quantity",
    aggfunc = "sum"
)
print(pivot_table1)
```

Create a Pivot Table showing Average Revenue for each Store in every Country.

```python
pivot_table2 = master_sales.pivot_table(
    index=["country", "store_name"],
    values="revenue",
    aggfunc="mean"
)

print(pivot_table2) 
```

Create a Crosstab showing Warranty Status vs Category

```python
cross_table = pd.crosstab(
    index=master_sales["claim_status"],
    columns=master_sales["category_name"]
)

print(cross_table)
```

Create a Dashboard-Ready Pivot Table containing Country, Category, Revenue, Quantity Sold,
Average Quantity, and Warranty Claims

```python
dashboard_pivot = master_sales.pivot_table(
    index=["country", "category_name"],
    values=["revenue", "quantity", "claim_status"],
    aggfunc={
        "revenue": "sum",
        "quantity": ["sum", "mean"],
        "claim_status": "count"
    }
)

print(dashboard_pivot)
```
