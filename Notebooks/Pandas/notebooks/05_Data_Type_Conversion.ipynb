```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

Convert sale_date into DateTime format.

```python
sales["sale_date"] = pd.to_datetime(sales["sale_date"])
```

Convert launch_date into DateTime format.

```python
products["launch_date"] = pd.to_datetime(products["launch_date"])
```

```python
products.dtypes
```

Convert Country, Category, and Warranty Status into Category data type

```python
stores["country"] = stores["country"].astype("category")
```

```python
category["category_name"] = category["category_name"].astype("category")
```

```python
warranty["claim_status"] = warranty["claim_status"].astype("category")
```
