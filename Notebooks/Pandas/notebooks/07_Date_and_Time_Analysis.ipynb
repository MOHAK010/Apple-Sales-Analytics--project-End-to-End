```python

import pandas as pd
import numpy as np 
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

Extract the Year from the Sale Date column.

```python
sales["sale_date"] = pd.to_datetime(sales["sale_date"])
products["launch_date"] = pd.to_datetime(products["launch_date"])
```

```python
sales["year"] = sales["sale_date"].dt.year
```

```python
print(sales[["sale_id","year"]])
```

Extract the Month Name from the Sale Date column

```python
sales["month"] = sales["sale_date"].dt.month_name()
```

```python
print(sales[["sale_id","month"]])
```

Extract the Quarter and Weekday from the Sale Date column

```python
sales["quarter"] = sales["sale_date"].dt.quarter

sales["weekday"] = sales["sale_date"].dt.day_name()
```

```python
print(sales[["sale_id", "sale_date", "quarter", "weekday"]])
```

Identify the month having the highest number of sales records.

```python
sales["month"].value_counts()
```

```python
sales["sale_date"].dt.month_name().value_counts().head(1)
```

Prepare a Monthly Sales Trend table containing Year, Month, Number of Sales, and Total Quantity
Sold.

```python
monthly_sales = (
    sales
    .groupby(["year", "month"])
    .agg(
        number_of_sales=("sale_id", "count"),
        total_quantity_sold=("quantity", "sum")
    )
    .reset_index()
)

monthly_sales
```
