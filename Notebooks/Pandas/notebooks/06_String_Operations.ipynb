```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

Convert all Product Names into uppercase.

```python
products["product_name"] = products["product_name"].str.upper()
```

Remove extra spaces from all text columns wherever applicable.

```python
stores["store_name"] = stores["store_name"].str.strip()

stores["city"] = stores["city"].str.strip()

stores["country"] = stores["country"].str.strip()
```

Find all products whose names contain the word "iPhone"

```python
products[
    products["product_name"].str.contains("iphone", case=False)
]

# Case = False because it will match any case sensitive word like "iPhone", "IPHONE" , "iphone"
```

Replace the word "IPHONE" with "APPLE" in the Product Name column.

```python
products["product_name"] = products["product_name"].str.replace("IPHONE","APPLE",case = False)
```

```python
print(products)
```

Create a searchable Product Catalog containing Product Name, Category, Price, and Launch Date
sorted alphabetically

```python
product_catalog = products.merge(
    category,
    on="category_id"
)

product_catalog = product_catalog[
    [
        "product_name",
        "category_name",
        "price",
        "launch_date"
    ]
]

product_catalog = product_catalog.sort_values(
    by="product_name"
)

product_catalog
```
