```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

# Prepare a filtered dataset showing Premium Apple products sold only through UK stores with complete store and product information.

```python
sales_products = sales.merge(
    products,
    on="product_id"
)

sales_products_store = sales_products.merge(
    stores,
    on="store_id"
)

premium_UK = sales_products_store[
    (sales_products_store["price"] > 1000)
    &
    (sales_products_store["country"] == "UK")
]

premium_UK
```
