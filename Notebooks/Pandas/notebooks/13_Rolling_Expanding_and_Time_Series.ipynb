```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
products = pd.read_csv("../../../Data/products.csv")
stores = pd.read_csv("../../../Data/stores.csv")
category = pd.read_csv("../../../Data/category.csv")
warranty = pd.read_csv("../../../Data/warranty.csv")
```

Calculate the Running Total Revenue based on Sale Date.

```python
master_sales = sales.merge(products, on="product_id")
master_sales = master_sales.merge(category, on="category_id")
master_sales = master_sales.merge(stores, on="store_id")
master_sales = master_sales.merge(warranty, on="sale_id")

# First make the master_sales DataFrame by merge all the dataframes together.
```

```python
master_sales["revenue"] = master_sales["quantity"] * master_sales["price"]

# Calculate the Revenue Contribution of each store based on the total revenue.
```

```python
master_sales = master_sales.sort_values("sale_date")

# Calculate the Running Total Revenue based on Sale Date.
```

```python
master_sales["running_total_revenue"] = master_sales["revenue"].cumsum()
```

```python
master_sales[["sale_date", "revenue", "running_total_revenue"]].head(10)
```

Calculate a 7-Day Moving Average of Revenue

```python
master_sales["moving_average"] = master_sales["revenue"].rolling(window=7).mean()
```

```python
print(master_sales[["sale_date", "revenue", "moving_average"]].head(10))
```

Create a column showing the Previous Day Revenue using shift()

```python
master_sales["previous_revenue"] = master_sales["revenue"].shift(1)
```

```python
print(master_sales[["sale_date", "revenue", "previous_revenue"]].head(10))
```

Calculate the Revenue Growth (%) compared to the previous day

```python
master_sales["revenue_growth"] = (
    (master_sales["revenue"] - master_sales["previous_revenue"])
    / master_sales["previous_revenue"]
) * 100
```

```python
print(master_sales[["sale_date", "revenue", "previous_revenue", "revenue_growth"]].head(10))
```

Build a complete Sales Trend Report containing Date, Daily Revenue, Previous Revenue, Growth (%),
Running Revenue, and 7-Day Moving Average.

```python
sales_trend = master_sales[
    [
        "sale_date",
        "revenue",
        "previous_revenue",
        "revenue_growth",
        "running_total_revenue",
        "moving_average"
    ]
]

sales_trend
```
