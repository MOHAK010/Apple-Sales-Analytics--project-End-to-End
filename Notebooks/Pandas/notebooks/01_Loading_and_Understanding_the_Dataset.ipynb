```python
# Create a summary table showing the number of rows, columns, missing values, and memory usage for each dataset.
```

```python
print(sales.shape[0])
sales.shape[1]
```

```python
sales.isnull().sum().sum() 

# 1st sum()  -> Count missing values in each column
# 2nd sum()  -> Add all column counts to get total missing values
```

```python
sales.memory_usage()
```

```python
summary = pd.DataFrame()

# Add dataset names
summary["Dataset"] = ["Sales", "Products", "Stores", "Category", "Warranty"]

# Add total rows
summary["Rows"] = [
    sales.shape[0],
    products.shape[0],
    stores.shape[0],
    category.shape[0],
    warranty.shape[0]
]

# Add total columns
summary["Columns"] = [
    sales.shape[1],
    products.shape[1],
    stores.shape[1],
    category.shape[1],
    warranty.shape[1]
]

# Add total missing values
summary["Missing Values"] = [
    sales.isnull().sum().sum(),       # Column-wise Missing -> Total Missing
    products.isnull().sum().sum(),
    stores.isnull().sum().sum(),
    category.isnull().sum().sum(),
    warranty.isnull().sum().sum()
]

# Add total memory usage
summary["Memory Usage (Bytes)"] = [
    sales.memory_usage(deep=True).sum(),       # Memory of all columns
    products.memory_usage(deep=True).sum(),
    stores.memory_usage(deep=True).sum(),
    category.memory_usage(deep=True).sum(),
    warranty.memory_usage(deep=True).sum()
]

# Display final report
summary
```

```python
# Create a DataFrame report identifying each dataset as either a Fact Table or Dimension Table, along with its business purpose.
```

```python
table_info = pd.DataFrame()

# Table Name

table_info['Table'] = [
    "sales",
    "products",
    "stores",
    "category",
    "warranty"
]

# Table Type

table_info['Type'] = [
    "Fact",
    "Dimension",
    "Dimension",
    "Dimension",
    "Dimension"
]

# Table Purpose

table_info["Purpose"] = [
    "Stores all sales transactions",
    "Stores product details",
    "Stores store information",
    "Stores product category details",
    "Stores warranty claim information"
]

# Display report 

table_info
```

```python
# Prepare a professional Dataset Overview Report containing dataset name, row count, column count, primary key, foreign keys, and business purpose of each table.
```

```python
print(sales.columns)
print(stores.columns)
print(warranty.columns)
print(products.columns)
print(category.columns)
```

```python
report = pd.DataFrame()

# Add dataset names
report["Dataset"] = [
    "Sales",
    "Products",
    "Stores",
    "Category",
    "Warranty"
]

# Add total rows
report["Rows"] = [
    sales.shape[0],
    products.shape[0],
    stores.shape[0],
    category.shape[0],
    warranty.shape[0]
]

# Add total columns
report["Columns"] = [
    sales.shape[1],
    products.shape[1],
    stores.shape[1],
    category.shape[1],
    warranty.shape[1]
]

# Add Primary Key
report["Primary Key"] = [
    "sale_id",
    "product_id",
    "store_id",
    "category_id",
    "claim_id"
]

# Add Foreign Keys
report["Foreign Keys"] = [
    "product_id, store_id",
    "category_id",
    "-",
    "-",
    "sale_id"
]

# Add business purpose
report["Purpose"] = [
    "Stores all sales transactions",
    "Stores product details",
    "Stores store details",
    "Stores product category details",
    "Stores warranty claim details"
]

# Display final report
report
```

```python
# Business Insight:
# The Sales table is the central transaction (Fact) table, while Products,
# Stores, Category, and Warranty are Dimension tables that enrich sales data
# for business reporting and analysis.
```
