```python
import numpy as np 
import pandas as pd
```

```python
sales = pd.read_csv("../../../Data/sales.csv")
```

```python
quantity = np.array(sales['quantity'])
```

```python
# Calculate the total quantity sold using np.sum()
```

```python
print(np.sum(quantity))
```

```python
# Calculate the Average Quantity Sold
```

```python
print(np.average(quantity))
print(np.mean(quantity))   # Same but np.mean() is more commonly used for calculating the mean of an array.
```

```python
# Find the Minimum and Maximum Quantity sold
```

```python
print(f"Minimum quantity sold: {np.min(quantity)}")
print(f"Maximum quantity sold: {np.max(quantity)}")
```

```python
print(np.min(quantity))
print(np.max(quantity))
```

```python
# Using a reshaped Quantity matrix, calculate row-wise and column-wise totals
```

```python
quantity_matrix = quantity[:100].reshape(10,10)
print(quantity_matrix)
```

```python
row_totals = np.sum(quantity_matrix, axis=1)
col_total = np.sum(quantity_matrix, axis =0)
```

```python
print(row_totals)
print(col_total)
```

```python
# Build a Sales Summary Report containing Total Quantity, Average Quantity, Maximum Quantity, Minimum Quantity, and explain which KPI would be most useful for a business manager.
```

```python
print(f"Total Quantity Sold: {np.sum(quantity)}")
print(f"Average Quantity Sold: {np.mean(quantity)}")
print(f"Maximum Quantity Sold: {np.max(quantity)}")
print(f"Minimum Quantity Sold: {np.min(quantity)}")

print("\nMost useful KPI: Average Quantity Sold")
print("Reason: It shows the typical quantity sold per sale.")
```
