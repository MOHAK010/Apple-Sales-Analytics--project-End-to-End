```python
import pandas as pd 
import numpy as np 
```

```python
sales = pd.read_csv("../../../Data/Sales.csv")
```

```python
quantity = sales['quantity'].to_numpy()
```

```python
# Calculate the Mean Quantity sold.
```

```python
np.mean(quantity)
```

```python
# Calculate the Median Quantity sold.
```

```python
np.median(quantity)
```

```python
# Calculate the Variance and Standard Deviation of the Quantity values.
```

```python
print(np.std(quantity)) # Standard Deviation
print(np.var(quantity)) # Variance
```

```python
# Calculate the 25th, 50th, and 75th Percentiles of Quantity sold and interpret what they represent.
```

```python
np.percentile(quantity, 25), np.percentile(quantity, 50), np.percentile(quantity, 75)  # it ha many ways to write this code. You can also write it as below.
```

```python
print(np.unique(quantity, return_counts=True))

# The dataset is heavily concentrated around single-unit sales. Since most transactions have a quantity of 1, the 25th, 50th, and 75th percentiles are all 1.
```

```python
# Normalize the Quantity values between 0 and 1 and identify potential outliers using statistical measures .
```

```python
normalized_quantity = (quantity - np.min(quantity)) / (np.max(quantity) - np.min(quantity))
```

```python
print(normalized_quantity[-20:])

# Business Insight:
# Normalization removes the effect of different value scales, allowing fair comparison
# between sales records and improving the accuracy of further analysis.
```

```python
print(np.min(quantity))
print(np.max(quantity))
```

```python
str = np.std(quantity)
mean = np.mean(quantity)
```

```python
lower_limit = mean - 3 * str
upper_limit = mean + 3 * str 
```

```python
print(lower_limit, upper_limit)
```

```python
outliers = quantity[(quantity < lower_limit) | (quantity > upper_limit)]
```

```python
print(outliers)
```

```python
# Business Insight:
# Potential outliers were identified using the Mean ± 3×Standard Deviation rule.
# These values should be reviewed as they may represent unusual sales or bulk purchases.
```
