```python
# Rank all sales based on Quantity using argsort() without modifying the original dataset and identify the Top 20 highest quantity sales
```

```python
rank = np.argsort(quantity)
```

```python
top_20_indexes = rank[-20:][::-1]    # We can use the negative indexing to get the last 20 elements and then reverse the order to get the top 20 highest quantity sales
```

```python
top_20_quantity = quantity[top_20_indexes]
```

```python
print("Top 20 highest quantity sales are: ", top_20_quantity)
```
