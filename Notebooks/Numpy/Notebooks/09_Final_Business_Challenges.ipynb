```python
import numpy as np 
import pandas as pd
```

```python
sales = pd.read_csv('../../../Data/Sales.csv')
```

```python
quantity = sales['quantity'].to_numpy()
```

```python
# Identify all sales where the Quantity sold is above the overall average Quantity
```

```python
average_quantity = quantity.mean()
```

```python
print(average_quantity)
```

```python
average_quantity_sales_and_above = quantity[quantity >= average_quantity]
```

```python
print(average_quantity_sales_and_above)
```

```python
# Display the Top 20 highest Quantity sales using NumPy only
```

```python
rank = np.argsort(quantity)
```

```python
top_20 = rank[-20:][::-1]
```

```python
sales.iloc[top_20]
```

```python
# Calculate the contribution percentage of every sale to the total Quantity sold.
```

```python
total_quantity = quantity.sum()
print(total_quantity)
```

```python
percentage_contribiution = np.round((quantity / total_quantity) * 100, 4)
print(percentage_contribiution)
```

```python
# Simulate a 15% increase in product demand and calculate the updated total Quantity sold.
```

```python
updated_quantity = total_quantity * 1.15
```

```python
print(updated_quantity)
```
